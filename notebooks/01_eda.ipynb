# EDA — NEU Surface Defect Database

Goals of this notebook:
1. Confirm the dataset is placed correctly (`data/raw/NEU-DET/...`)
2. Check class balance
3. Visualize sample images per class
4. Inspect pixel intensity distributions
5. Preview the OpenCV preprocessing pipeline (resize, edges, contours)

```python
import sys
sys.path.append('..')

import matplotlib.pyplot as plt
from src.data.dataset import NEUDataset, CLASS_NAMES
from src.utils.preprocessing import load_grayscale, extract_edges, extract_contours

DATA_ROOT = '../data/raw/NEU-DET'  # adjust if your layout has train/validation subfolders
```

```python
dataset = NEUDataset(DATA_ROOT)
print(f'Total images: {len(dataset)}')
dataset.class_distribution()
```

```python
# Class balance bar chart
dist = dataset.class_distribution()
plt.figure(figsize=(8, 4))
plt.bar(dist.keys(), dist.values())
plt.xticks(rotation=45, ha='right')
plt.ylabel('Count')
plt.title('NEU-DET class distribution')
plt.tight_layout()
plt.show()
```

```python
# One sample image per class
fig, axes = plt.subplots(1, len(CLASS_NAMES), figsize=(18, 3))
for ax, class_name in zip(axes, CLASS_NAMES):
    sample_path = next(p for p, l in dataset.samples if CLASS_NAMES[l] == class_name)
    img = load_grayscale(sample_path)
    ax.imshow(img, cmap='gray')
    ax.set_title(class_name, fontsize=9)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Preview edges + contours on one example (shared toolkit with the thesis pipeline)
sample_path = dataset.samples[0][0]
img = load_grayscale(sample_path)
edges = extract_edges(img)
contours = extract_contours(img)

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(img, cmap='gray'); axes[0].set_title('Original'); axes[0].axis('off')
axes[1].imshow(edges, cmap='gray'); axes[1].set_title('Canny edges'); axes[1].axis('off')
plt.tight_layout()
plt.show()

print(f'Largest contour area: {__import__("cv2").contourArea(contours[0]) if contours else "none found"}')
```
